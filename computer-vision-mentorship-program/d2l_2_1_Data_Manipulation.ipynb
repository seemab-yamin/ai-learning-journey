nn.Conv2d

```python
import torch
import torch.nn as nn
```

```python
conv = nn.Conv2d(1, 1, kernel_size=3)

image = torch.randn(1, 1, 28, 28)
```

```python
output = conv(image)
print(image.shape, output.shape)
```

```python
# adding padding to keep size same
conv = nn.Conv2d(1, 1, kernel_size=3, padding=1)
```

```python
output = conv(image)
print(image.shape, output.shape)
```

```python
# with stride 2 we jump two pixels at a time so output reduced to half
conv = nn.Conv2d(1, 1, kernel_size=3, padding=1, stride=2)
```

```python
output = conv(image)
print(image.shape, output.shape)
```

```python
# output channel 32 means that we will have 32 different filters applied to the input image, 
# resulting in 32 different feature maps in the output. 
# Each filter will learn to detect different features from the input image, 
# allowing the model to capture more complex patterns and representations.
conv_multi = nn.Conv2d(1, 32, kernel_size=3, padding=1)
```

```python
output = conv_multi(image)
print(image.shape, output.shape)
```

```python
color_image = torch.randn(1, 3, 28, 28)
```

```python
conv_color = nn.Conv2d(3, 32, kernel_size=3, padding=1)
```

```python
output = conv_color(color_image)
print(color_image.shape, output.shape)
```

```python
depthwise_conv = nn.Conv2d(2, 4, padding=1, kernel_size=3, groups=2)
```

```python
color_image = torch.randn(1, 2, 10, 10)
output = depthwise_conv(color_image)
print(color_image.shape, output.shape)
```
